# Results notebook

Thin wrapper over `eval.analyze` (no analysis logic lives here). Point `EXPERIMENTS` at the four **frozen** experiment directories after scoring. Dev-set experiments are for tuning only and are flagged in the warnings.

```python
from pathlib import Path

from IPython.display import Image, Markdown, display

from eval.analyze import analyze, load_experiment, render_markdown, write_outputs

EXPERIMENTS = [Path(f"out/experiments/{v}-frozen") for v in ("a0", "a1", "a2", "a3")]
OUT = Path("analysis")
```

```python
an = analyze([load_experiment(p) for p in EXPERIMENTS if p.exists()])
write_outputs(an, OUT)
display(Markdown(render_markdown(an)))
```

## Exploratory pairwise and per-seed tests

```python
import csv

for name in ("pairwise_exploratory.csv", "per_seed_tests.csv"):
    with (OUT / name).open() as fh:
        rows = list(csv.DictReader(fh))
    lines = [
        f"- {r['comparison']} (seed {r['seed'] or 'majority'}): "
        f"diff={float(r['diff']):+.3f}, p={float(r['p_value']):.4f}"
        for r in rows
    ]
    display(Markdown(f"### {name}\n" + "\n".join(lines)))
```
